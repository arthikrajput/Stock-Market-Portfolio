```python
from flask import Flask, render_template_string, request
import pandas as pd
import plotly.graph_objects as go
from plotly.io import to_html

# Load datasets
fundamentals_df = pd.read_csv("fundamentals.csv")
prices_df = pd.read_csv("prices.csv")
securities_df = pd.read_csv("securities.csv")

# Merge datasets
merged_fundamentals = pd.merge(
    fundamentals_df,
    securities_df,
    left_on='Ticker Symbol',
    right_on='Ticker symbol',
    how='inner'
)

# Calculate P/E and P/B Ratios
merged_fundamentals['P/E Ratio'] = merged_fundamentals['Total Revenue'] / merged_fundamentals['Net Income']
merged_fundamentals['P/B Ratio'] = merged_fundamentals['Total Revenue'] / merged_fundamentals['Total Equity']
merged_fundamentals['Period Ending'] = pd.to_datetime(merged_fundamentals['Period Ending'])

# Initialize Flask app
app = Flask(__name__)

@app.route("/", methods=["GET", "POST"])
def index():
    fig_html = ""
    selected_stock = ""
    
    if request.method == "POST":
        # Get stock ticker from user input
        selected_stock = request.form.get("stock").upper()

        # Filter data for the user-defined stock ticker
        stock_data = merged_fundamentals[merged_fundamentals['Ticker Symbol'] == selected_stock]
        if stock_data.empty:
            fig_html = f"<p>No data found for ticker {selected_stock}</p>"
        else:
            stock_data = stock_data[['Period Ending', 'P/E Ratio', 'P/B Ratio']].dropna()
            stock_data = stock_data.sort_values(by='Period Ending')

            # Create the Plotly graph
            fig = go.Figure()
            fig.add_trace(go.Scatter(
                x=stock_data['Period Ending'], 
                y=stock_data['P/E Ratio'], 
                mode='lines+markers', 
                name='P/E Ratio', 
                line=dict(color='blue')
            ))
            fig.add_trace(go.Scatter(
                x=stock_data['Period Ending'], 
                y=stock_data['P/B Ratio'], 
                mode='lines+markers', 
                name='P/B Ratio', 
                line=dict(color='orange')
            ))

            # Update layout
            fig.update_layout(
                title=f'P/E and P/B Ratios Over Time for {selected_stock}',
                xaxis_title='Period Ending',
                yaxis_title='Ratio',
                template='plotly_white'
            )

            # Convert Plotly figure to HTML
            fig_html = to_html(fig, full_html=False)

    # Render the form and graph
    return render_template_string("""
    <!DOCTYPE html>
    <html>
    <head>
        <title>Flask Plotly Integration</title>
    </head>
    <body>
        <h1>Enter Stock Ticker</h1>
        <form method="POST">
            <label for="stock">Stock Ticker:</label>
            <input type="text" id="stock" name="stock" placeholder="e.g., AAPL">
            <button type="submit">Submit</button>
        </form>
        <div>{{ graph|safe }}</div>
    </body>
    </html>
    """, graph=fig_html)

if __name__ == "__main__":
    app.run(debug=True)
```
